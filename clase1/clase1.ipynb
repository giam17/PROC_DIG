con colores relleno

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [10,5]
img = cv2.imread('v4033_197.png', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = imgRGB[:,:,2]
g = imgRGB[:,:,1]
b = imgRGB[:,:,0]

#crear imagen por canal
r_img = np.zeros_like(imgRGB)
r_img[:,:,0] = r

g_img = np.zeros_like(imgRGB)
g_img[:,:,1] = g

b_img = np.zeros_like(imgRGB)
b_img[:,:,2] = b

plt.figure(figsize=(15,5))

plt.subplot(141)
plt.title('Imagen Original')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(142)
plt.title('Canal Rojo')
plt.imshow(r_img)
plt.axis('off')

plt.subplot(143)
plt.title('Canal Verde')
plt.imshow(g_img)
plt.axis('off')

plt.subplot(144)
plt.title('Canal Azul')
plt.imshow(b_img)
plt.axis('off')

plt.show()
```

generacion de falso color con cmap

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [10,5]
img = cv2.imread('v4033_197.png', cv2.IMREAD_COLOR)
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r = imgRGB[:,:,2]
g = imgRGB[:,:,1]
b = imgRGB[:,:,0]

#crear imagen por canal
r_img = np.zeros_like(imgRGB)
r_img[:,:,0] = r

g_img = np.zeros_like(imgRGB)
g_img[:,:,1] = g

b_img = np.zeros_like(imgRGB)
b_img[:,:,2] = b


gray = cv2.cvtColor(imgRGB, cv2.COLOR_RGB2GRAY)

imgHSV = cv2.cvtColor(imgRGB, cv2.COLOR_RGB2HSV)

plt.figure(figsize=(24,8))

plt.subplot(2, 6, 1)
plt.title('Imagen Original')
plt.imshow(imgRGB)
plt.axis('off')

plt.subplot(2, 6, 2)
plt.title('Canal Rojo')
plt.imshow(r_img)
plt.axis('off')

plt.subplot(2, 6, 3)
plt.title('Canal Verde')
plt.imshow(g_img)
plt.axis('off')

plt.subplot(2, 6, 4)
plt.title('Canal Azul')
plt.imshow(b_img)
plt.axis('off')

plt.subplot(2, 6, 5)
plt.title('Escala de Grises')
plt.imshow(gray, cmap='gray')
plt.axis('off')

plt.subplot(2, 6, 6)
plt.title('Canal R (Reds)')
plt.imshow(r, cmap='Reds')
plt.axis('off')

# Fila 2 - 5 imágenes
plt.subplot(2, 6, 7)
plt.title('Canal G (Greens)')
plt.imshow(g, cmap='Greens')
plt.axis('off')

plt.subplot(2, 6, 8)
plt.title('Canal B (Blues)')
plt.imshow(b, cmap='Blues')
plt.axis('off')

plt.subplot(2, 6, 9)
plt.title('Matiz (Hue)')
plt.imshow(imgHSV[:,:,0], cmap='hsv')
plt.axis('off')

plt.subplot(2, 6, 10)
plt.title('Saturación')
plt.imshow(imgHSV[:,:,1], cmap='gray')
plt.axis('off')

plt.subplot(2, 6, 11)
plt.title('Valor (Value)')
plt.imshow(imgHSV[:,:,2], cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```
