```python
import matplotlib.pyplot as plt
import numpy as np

width = 100
height = 100
image = np.zeros((height, width, 3), dtype=np.uint8)

for y in range(height):
    for x in range(width):
        if y < height // 2 and x < width // 2:
            image[y][x] = [255, 0, 0]
        elif y < height // 2 and x >= width // 2:
            image[y][x] = [0, 255, 0]
        elif y >= height // 2 and x < width // 2:
            image[y][x] = [0, 0, 255]
        else:
            image[y][x] = [0, 0, 0]

plt.imshow(image)
plt.title('Cuadrandtes rojo, verde, azul, negro')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import random

width = 150
height = 150
image = np.zeros((height, width, 3), dtype=np.uint8)

colores_base = [
    [255, 0, 0],
    [0, 255, 0],
    [255, 255, 255],
    [255, 165, 0],
    [255, 255, 0],
    [0, 0, 255],
]

colores = (colores_base * 2)[:9]
random.shuffle(colores)

tam = height // 3

for y in range(height):
    for x in range(width):
        fila = y // tam
        col  = x // tam
        indice = fila * 3 + col
        image[y][x] = colores[indice]

plt.imshow(image)
plt.axis('off')
plt.title('Cubo Rubik 3x3 - 6 Colores')
plt.show()
```
