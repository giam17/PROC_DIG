```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def mostrar_imagen(img_path, titulo):
    img_bgr = cv2.imread(img_path, cv2.IMREAD_COLOR)
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

    r_img = np.zeros_like(img_rgb)
    r_img[:, :, 0] = img_rgb[:, :, 0]

    g_img = np.zeros_like(img_rgb)
    g_img[:, :, 1] = img_rgb[:, :, 1]

    b_img = np.zeros_like(img_rgb)
    b_img[:, :, 2] = img_rgb[:, :, 2]

    cv2.imwrite(img_path.replace('.jpeg', '_bgr.jpeg'), img_bgr)
    cv2.imwrite(img_path.replace('.jpeg', '_gray.jpeg'), img_gray)

    fig, axs = plt.subplots(1, 6, figsize=(24, 4))
    fig.suptitle(titulo, fontsize=14)

    axs[0].imshow(img_rgb);
    axs[0].set_title('Original (RGB)');
    axs[0].axis('off')
    axs[1].imshow(r_img);
    axs[1].set_title('Canal Rojo');
    axs[1].axis('off')
    axs[2].imshow(g_img);
    axs[2].set_title('Canal Verde');
    axs[2].axis('off')
    axs[3].imshow(b_img);
    axs[3].set_title('Canal Azul');
    axs[3].axis('off')
    axs[4].imshow(img_gray, cmap='gray');
    axs[4].set_title('Escala de grises');
    axs[4].axis('off')
    axs[5].imshow(img_gray, cmap='hot');
    axs[5].set_title('Color falso');
    axs[5].axis('off')

    plt.tight_layout()
    plt.show()

mostrar_imagen('imagen1.jpeg', 'Imagen 1')
```
