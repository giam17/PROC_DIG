```python
import cv2
import numpy as np

img = cv2.imread('giam.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

laplaciano = cv2.Laplacian(gray, cv2.CV_64F)
laplaciano_abs = cv2.convertScaleAbs(laplaciano)

blur = cv2.GaussianBlur(gray, (5, 5), 0)

log = cv2.Laplacian(blur, cv2.CV_64F)
log_abs = cv2.convertScaleAbs(log)

cv2.imwrite('resultado_laplaciano.jpg', laplaciano_abs)
cv2.imwrite('resultado_log.jpg', log_abs)

cv2.imshow('Imagen Original', img)
cv2.imshow('Filtro Laplaciano', laplaciano_abs)
cv2.imshow('Filtro LoG (Laplacian of Gaussian)', log_abs)
```
