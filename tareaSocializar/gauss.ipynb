```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('giam.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

laplaciano = cv2.Laplacian(gray, cv2.CV_64F)
laplaciano_abs = cv2.convertScaleAbs(laplaciano)

blur = cv2.GaussianBlur(gray, (5, 5), 0)
log = cv2.Laplacian(blur, cv2.CV_64F)
log_abs = cv2.convertScaleAbs(log)

cv2.imwrite('resultado_laplaciano.jpg', laplaciano_abs)
cv2.imwrite('resultado_log.jpg', log_abs)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(gray, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Laplaciano")
plt.imshow(laplaciano_abs, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("LoG laplaciano gaussiano")
plt.imshow(log_abs, cmap="gray")
plt.axis("off")

plt.tight_layout()
plt.show()
```
