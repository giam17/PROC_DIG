```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

width, height = 800, 800
img_diana = np.ones((height, width), dtype=np.uint8) * 255

center = (width // 2, height // 2)
radios = [400, 320, 240, 160, 80]
colores = [0, 255, 0, 255, 0]

for r, color in zip(radios, colores):
    cv2.circle(img_diana, center, r, color, -1)

plt.figure(figsize=(6, 6))
plt.imshow(img_diana, cmap='gray', vmin=0, vmax=255)
plt.title('Ejercicio 1: Diana de tiro al blanco')
plt.axis('on')
plt.show()
```
