```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((1, tkernel), np.float32) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_v(imagen, tkernel):
    kernel = np.ones((tkernel, 1), np.float32) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_d(imagen, tkernel):
    kernel = np.eye(tkernel, dtype=np.float32) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_di(imagen, tkernel):
    kernel = np.fliplr(np.eye(tkernel, dtype=np.float32)) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

img = cv2.imread('imagen3.jpg')

if img is None:
    print("Error: No se pudo cargar la imagen. Verifica que el archivo exista en la ruta.")
else:
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    tkernel = 45 

    img_h = convolucion_media_h(img, tkernel)
    img_v = convolucion_media_v(img, tkernel)
    img_d = convolucion_media_d(img, tkernel)
    img_di = convolucion_media_di(img, tkernel)

    plt.figure(figsize=(15, 12))

    plt.subplot(3, 2, 1)
    plt.imshow(img)
    plt.title('Original')
    plt.xticks([]), plt.yticks([])

    plt.subplot(3, 2, 2)
    plt.imshow(img_h)
    plt.title('Convolucionada Horizontal')
    plt.xticks([]), plt.yticks([])

    plt.subplot(3, 2, 3)
    plt.imshow(img_v)
    plt.title('Convolucionada Vertical')
    plt.xticks([]), plt.yticks([])

    plt.subplot(3, 2, 4)
    plt.imshow(img_d)
    plt.title('Convolucionada Diagonal')
    plt.xticks([]), plt.yticks([])

    plt.subplot(3, 2, 5)
    plt.imshow(img_di)
    plt.title('Convolucionada Diagonal Inversa')
    plt.xticks([]), plt.yticks([])

    plt.subplot(3, 2, 6)
    plt.imshow(img)
    plt.title('Original de Nuevo')
    plt.xticks([]), plt.yticks([])

    plt.tight_layout()
    plt.show()
```
