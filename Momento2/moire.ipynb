```python
# Contrarrestar Moiré - Imagen 1
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img1 = cv2.imread('moire1.jpg')
img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)

kernel_size = (15, 15)
sigma = 3
img1_filtrada = cv2.GaussianBlur(img1, kernel_size, sigma)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(img1)
axes[0].set_title('Moiré Original - Imagen 1')
axes[0].axis('off')

axes[1].imshow(img1_filtrada)
axes[1].set_title('Moiré Contrarrestado - Imagen 1')
axes[1].axis('off')

plt.suptitle('Sección 1: moire1', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
