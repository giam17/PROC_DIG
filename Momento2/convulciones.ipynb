```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def aplicar_filtros_y_mostrar(ruta_imagen):
    img_bgr = cv2.imread(ruta_imagen)
    if img_bgr is None:
        print(f"Error: No se pudo cargar la imagen {ruta_imagen}")
        return

    img_color = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    img_gris = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

    kernels = {
        "Desenfoque": np.array([
            [1, 1, 1],
            [1, 1, 1],
            [1, 1, 1]
        ]) / 9.0,

        "Realce de bordes": np.array([
            [0, 0, 0],
            [-1, 1, 0],
            [0, 0, 0]
        ]),

        "Repujado": np.array([
            [-2, -1, 0],
            [-1, 1, 1],
            [0, 1, 2]
        ]),

        "Detección de bordes": np.array([
            [0, 1, 0],
            [1, -4, 1],
            [0, 1, 0]
        ]),

        "Filtro de tipo Sobel": np.array([
            [-1, 0, 1],
            [-2, 0, 2],
            [-1, 0, 1]
        ]),

        "Filtro de tipo Sharpen": np.array([
            [1, -2, 1],
            [-2, 5, -2],
            [1, -2, 1]
        ]),

        "Filtro Norte": np.array([
            [1, 1, 1],
            [1, -2, 1],
            [-1, -1, -1]
        ]),

        "Filtro Este": np.array([
            [-1, 1, 1],
            [-1, -2, 1],
            [-1, 1, 1]
        ]),

        "Filtro de tipo Gauss": np.array([
            [1, 2, 3, 2, 1],
            [2, 7, 11, 7, 2],
            [3, 11, 17, 11, 3],
            [2, 7, 11, 7, 2],
            [1, 2, 3, 2, 1]
        ]) / 115.0
    }

    fig, axs = plt.subplots(len(kernels), 3, figsize=(15, 4 * len(kernels)))

    for i, (nombre, kernel) in enumerate(kernels.items()):
        res_gris = cv2.filter2D(img_gris, -1, kernel)
        res_color = cv2.filter2D(img_color, -1, kernel)

        axs[i, 0].imshow(img_color)
        axs[i, 0].set_title(f"Original a Color")
        axs[i, 0].axis('off')

        axs[i, 1].imshow(res_gris, cmap='gray')
        axs[i, 1].set_title(f"{nombre} (Grises)")
        axs[i, 1].axis('off')

        axs[i, 2].imshow(res_color)
        axs[i, 2].set_title(f"{nombre} (Color)")
        axs[i, 2].axis('off')

    plt.tight_layout()
    plt.show()

aplicar_filtros_y_mostrar('imagen1.jpg')
```
