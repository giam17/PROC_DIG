```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def gaussian_kernel(size=15, sigma=3):
    k = size // 2
    kernel = np.zeros((size, size))
    for x in range(-k, k + 1):
        for y in range(-k, k + 1):
            kernel[x + k][y + k] = np.exp(-(x**2 + y**2) / (2 * sigma**2))
    kernel = kernel / kernel.sum()
    return kernel

def aplicar_filtro(imagen, kernel):
    pad = kernel.shape[0] // 2
    resultado = np.zeros_like(imagen, dtype=np.float64)
    img_padded = np.pad(imagen, ((pad, pad), (pad, pad), (0, 0)), mode='reflect')
    for i in range(imagen.shape[0]):
        for j in range(imagen.shape[1]):
            bloque = img_padded[i:i + kernel.shape[0], j:j + kernel.shape[1], :]
            resultado[i, j, :] = np.sum(bloque * kernel[:, :, np.newaxis], axis=(0, 1))
    return np.clip(resultado, 0, 255).astype(np.uint8)

kernel = gaussian_kernel(size=15, sigma=3)

img1 = np.array(Image.open('moire1.jpg').convert('RGB'))
img1_filtrada = aplicar_filtro(img1, kernel)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(img1)
axes[0].set_title('Moiré Original - Imagen 1')
axes[0].axis('off')
axes[1].imshow(img1_filtrada)
axes[1].set_title('Moiré Contrarrestado - Imagen 1')
axes[1].axis('off')
plt.suptitle('Sección 1: moire1', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
