```python
limite = int(input("¿Cuántos números desea ingresar? (mínimo 3): "))

while limite < 3:
    print("Error: Debe ingresar al menos 3 números")
    limite = int(input("¿Cuántos números desea ingresar? (mínimo 3): "))

numeros = []
for i in range(limite):
    numero = int(input(f"Ingrese el número {i + 1}: "))
    numeros.append(numero)

mayor = max(numeros)
menor = min(numeros)

print(f"El número mayor es: {mayor}")
print(f"El número menor es: {menor}")
```
