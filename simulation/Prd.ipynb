EJERCICIO #3

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

tam = 100
filas = 8
cols = 8

tablero = np.zeros((filas * tam, cols * tam), dtype=np.uint8)

for f in range(filas):
    for c in range(cols):
        y = f * tam
        x = c * tam
        if (f + c) % 2 == 0:
            tablero[y:y+tam, x:x+tam] = 255  
        else:
            tablero[y:y+tam, x:x+tam] = 0    

cv2.imwrite('tablero_ajedrez.png', tablero)

plt.imshow(tablero, cmap='gray')
plt.title('Tablero de Ajedrez')
plt.show()
```

EJERCICIO #4

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('degradado_lineal.png')
img_gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

def posterizar(imagen, niveles):
    factor_escala = 255 / niveles
    img_pos = np.round(imagen / factor_escala) * factor_escala
    return img_pos.astype(np.uint8)

def solarizar(imagen, umbral):
    img_sol = np.where(imagen < umbral, imagen, 255 - imagen)
    return img_sol.astype(np.uint8)

img_posterizada = posterizar(img_gray, 3)
img_solarizada  = solarizar(img_gray, 127)

cv2.imwrite('degradado_posterizado.png', img_posterizada)
cv2.imwrite('degradado_solarizado.png',  img_solarizada)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))

plt.subplot(1, 3, 2)
plt.title('Imagen Posterizada')
plt.imshow(img_posterizada, cmap='gray')

plt.subplot(1, 3, 3)
plt.title('Imagen Solarizada')
plt.imshow(img_solarizada, cmap='gray')

plt.tight_layout()
plt.savefig('poster_solarizado.png')
plt.show()

paletas = {
    'AUTUMN':  cv2.COLORMAP_AUTUMN,
    'BONE':    cv2.COLORMAP_BONE,
    'JET':     cv2.COLORMAP_JET,
    'WINTER':  cv2.COLORMAP_WINTER,
    'RAINBOW': cv2.COLORMAP_RAINBOW,
    'OCEAN':   cv2.COLORMAP_OCEAN,
    'SUMMER':  cv2.COLORMAP_SUMMER,
    'SPRING':  cv2.COLORMAP_SPRING,
    'COOL':    cv2.COLORMAP_COOL,
    'HSV':     cv2.COLORMAP_HSV,
    'HOT':     cv2.COLORMAP_HOT,
    'VIRIDIS': cv2.COLORMAP_VIRIDIS,
    'PLASMA':  cv2.COLORMAP_PLASMA,
    'INFERNO': cv2.COLORMAP_INFERNO,
    'MAGMA':   cv2.COLORMAP_MAGMA,
}

for nombre, colormap in paletas.items():
    resultado = cv2.applyColorMap(img_gray, colormap)
    cv2.imwrite(f'falso_color_{nombre}.png', resultado)  # solo guarda
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imagen1.jpg')
img_gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

alto, ancho = img_gray.shape
degradado = np.linspace(1, 0, alto).reshape(alto, 1) 
img_degradada = (img_gray * degradado).astype(np.uint8)

cv2.imwrite('imagen1_degradado.png', img_degradada)

fig, axs = plt.subplots(1, 2, figsize=(14, 6))
axs[0].imshow(img_gray,      cmap='gray'); axs[0].set_title('Imagen Original')
axs[1].imshow(img_degradada, cmap='gray'); axs[1].set_title('Imagen con Degradado Lineal')
plt.tight_layout()
plt.savefig('ejercicio5_resultado.png')
plt.show()
```
